# 🧑‍🔬 Human-in-the-Loop Face Recognition Pipeline

This notebook consolidates the entire face recognition pipeline into a single, executable document.

**Pipeline Stages:**
1. **Environment Setup** — Install and import dependencies
2. **Face Detection & Embedding** — Detect faces and extract 512-D embeddings (InsightFace / buffalo_l)
3. **Media Ingestion** — Process bulk images and sample video frames
4. **Unknown Queue Management** — Cluster unrecognized faces and promote labeled clusters
5. **Recognition Logger** — Log inference results to CSV
6. **Training** — Train an SVM/KNN classifier on known face embeddings
7. **Inference** — Predict identity from a new image or video
8. **Evaluation** — Measure accuracy, precision, recall, F1, and Unknown Rejection Rate
9. **Visualization** — Confusion matrix, similarity distributions, per-class metrics

## 2. Configuration

Edit the paths and thresholds below to match your setup.

````python
# ── Paths ──────────────────────────────────────────────
KNOWN_DIR       = "data/known_faces"       # Subfolders per person with reference images
QUEUE_DIR       = "data/unknown_queue"      # Auto-clustered unrecognized faces
TEST_DIR        = "data/test_set"           # Evaluation test set (subfolders per person + Unknown/)
MODEL_PATH      = "models/classifier.pkl"
EMBEDDINGS_PATH = "data/embeddings/face_embeddings.pkl"
LOG_PATH        = "data/logs/recognition_logs.csv"
PLOT_DIR        = "data/logs/plots"

# ── Thresholds ─────────────────────────────────────────
SIM_THRESHOLD      = 0.55    # Cosine similarity cutoff for "Known" vs "Unknown"
BLUR_THRESHOLD     = 15.0    # Laplacian variance below this → blurry → skip
MIN_FACE_SIZE      = 60      # Pixels; ignore tiny background faces
DET_SIZE           = (640, 640)
SAMPLE_FPS         = 1.5     # Frames per second sampled from video
CLUSTER_THRESHOLD  = 0.72    # Cosine sim to merge into existing unknown cluster
DUPLICATE_THRESHOLD = 0.92   # Cosine sim to skip near-identical frames
````

## 3. Face Detection & Embedding (`FaceDetectorEmbedder`)

Wraps InsightFace's `buffalo_l` model to detect faces, filter out blurry / tiny crops, and return L2-normalized 512-D embeddings.

````python
class FaceDetectorEmbedder:
    def __init__(self, det_size=(640, 640), min_face_size=60, blur_threshold=15.0):
        self.app = FaceAnalysis(name="buffalo_l", providers=["CPUExecutionProvider"])
        self.app.prepare(ctx_id=0, det_size=det_size)
        self.min_face_size = min_face_size
        self.blur_threshold = blur_threshold

    def _is_blurry(self, image_crop: np.ndarray) -> bool:
        gray = cv2.cvtColor(image_crop, cv2.COLOR_BGR2GRAY)
        variance = cv2.Laplacian(gray, cv2.CV_64F).var()
        return variance < self.blur_threshold

    def extract_faces(self, frame: np.ndarray) -> list[dict]:
        """
        Detects faces in a BGR frame, filters low-quality crops,
        and returns a list of dicts with crop, bbox, det_score, and 512-D embedding.
        """
        if frame is None or frame.size == 0:
            return []

        h, w, _ = frame.shape
        faces = self.app.get(frame)
        results = []

        for face in faces:
            x1, y1, x2, y2 = face.bbox.astype(int)
            x1, y1 = max(0, x1), max(0, y1)
            x2, y2 = min(w, x2), min(h, y2)

            face_w = x2 - x1
            face_h = y2 - y1

            if face_w < self.min_face_size or face_h < self.min_face_size:
                continue

            # 10% padding for cleaner thumbnails
            pad_x = int(face_w * 0.1)
            pad_y = int(face_h * 0.1)
            cx1, cy1 = max(0, x1 - pad_x), max(0, y1 - pad_y)
            cx2, cy2 = min(w, x2 + pad_x), min(h, y2 + pad_y)

            crop = frame[cy1:cy2, cx1:cx2]
            if crop.size == 0 or self._is_blurry(crop):
                continue

            results.append({
                "crop": crop,
                "bbox": [int(x1), int(y1), int(x2), int(y2)],
                "det_score": float(face.det_score),
                "embedding": face.normed_embedding,  # L2-normalized 512-D
            })

        return results


# Instantiate detector
detector = FaceDetectorEmbedder(
    det_size=DET_SIZE,
    min_face_size=MIN_FACE_SIZE,
    blur_threshold=BLUR_THRESHOLD,
)
print(f"Detector ready (det_size={DET_SIZE}, blur_threshold={BLUR_THRESHOLD}) ✅")
````

## 4. Media Ingestion (`MediaIngestor`)

Handles bulk image lists and video files, sampling frames at a configurable FPS.

````python
class MediaIngestor:
    def __init__(self, detector: FaceDetectorEmbedder, sample_fps: float = 1.5):
        self.detector = detector
        self.sample_fps = sample_fps

    def process_images(self, image_paths: list[str]) -> list[dict]:
        """Processes a bulk list of image file paths."""
        extracted = []
        for path in image_paths:
            frame = cv2.imread(path)
            if frame is None:
                continue
            faces = self.detector.extract_faces(frame)
            for idx, face in enumerate(faces):
                face["source_type"] = "image"
                face["source_name"] = os.path.basename(path)
                face["timestamp_sec"] = None
                face["face_index"] = idx
                extracted.append(face)
        return extracted

    def process_video(self, video_path: str) -> list[dict]:
        """Samples frames from a video at `sample_fps` and extracts clean faces."""
        cap = cv2.VideoCapture(video_path)
        if not cap.isOpened():
            raise ValueError(f"Could not open video file: {video_path}")

        native_fps = cap.get(cv2.CAP_PROP_FPS) or 30.0
        frame_interval = max(1, int(round(native_fps / self.sample_fps)))

        extracted = []
        frame_idx = 0
        video_name = os.path.basename(video_path)

        while True:
            ret, frame = cap.read()
            if not ret:
                break
            if frame_idx % frame_interval == 0:
                timestamp_sec = round(frame_idx / native_fps, 2)
                faces = self.detector.extract_faces(frame)
                for idx, face in enumerate(faces):
                    face["source_type"] = "video"
                    face["source_name"] = video_name
                    face["timestamp_sec"] = timestamp_sec
                    face["face_index"] = idx
                    extracted.append(face)
            frame_idx += 1

        cap.release()
        return extracted


ingestor = MediaIngestor(detector=detector, sample_fps=SAMPLE_FPS)
print(f"Ingestor ready (sample_fps={SAMPLE_FPS}) ✅")
````

## 5. Unknown Queue Manager (`UnknownQueueManager`)

Automatically clusters unknown faces by cosine similarity. When a human labels a cluster, its crops are promoted to `known_faces/`.

````python
class UnknownQueueManager:
    def __init__(self, queue_dir="data/unknown_queue", known_dir="data/known_faces",
                 cluster_threshold=0.72, duplicate_threshold=0.92):
        self.queue_dir = queue_dir
        self.known_dir = known_dir
        self.cluster_threshold = cluster_threshold
        self.duplicate_threshold = duplicate_threshold
        os.makedirs(self.queue_dir, exist_ok=True)
        os.makedirs(self.known_dir, exist_ok=True)

    def _load_cluster_centroids(self) -> dict[str, np.ndarray]:
        """Loads average embeddings for existing unknown clusters."""
        centroids = {}
        for cluster_id in os.listdir(self.queue_dir):
            cluster_path = os.path.join(self.queue_dir, cluster_id)
            emb_file = os.path.join(cluster_path, "embeddings.npy")
            if os.path.isdir(cluster_path) and os.path.exists(emb_file):
                embs = np.load(emb_file)
                if len(embs) > 0:
                    mean_emb = np.mean(embs, axis=0)
                    centroids[cluster_id] = mean_emb / np.linalg.norm(mean_emb)
        return centroids

    def add_unknown_face(self, face_data: dict) -> str:
        """
        Groups an unrecognized face into an existing cluster if similar,
        skips near-identical frames, or creates a new cluster folder.
        """
        emb = face_data["embedding"].reshape(1, -1)
        centroids = self._load_cluster_centroids()

        best_cluster = None
        best_sim = -1.0

        for cid, centroid in centroids.items():
            sim = float(cosine_similarity(emb, centroid.reshape(1, -1))[0][0])
            if sim > best_sim:
                best_sim = sim
                best_cluster = cid

        if best_cluster and best_sim >= self.cluster_threshold:
            cluster_path = os.path.join(self.queue_dir, best_cluster)
            emb_file = os.path.join(cluster_path, "embeddings.npy")
            existing_embs = np.load(emb_file)

            max_frame_sim = float(np.max(cosine_similarity(emb, existing_embs)))
            if max_frame_sim >= self.duplicate_threshold:
                return f"skipped_duplicate:{best_cluster}"

            updated_embs = np.vstack([existing_embs, emb])
            np.save(emb_file, updated_embs)
            target_cluster = best_cluster
        else:
            target_cluster = f"person_{uuid.uuid4().hex[:6]}"
            cluster_path = os.path.join(self.queue_dir, target_cluster)
            os.makedirs(cluster_path, exist_ok=True)
            np.save(os.path.join(cluster_path, "embeddings.npy"), emb)

        img_name = f"{uuid.uuid4().hex[:8]}.jpg"
        cv2.imwrite(os.path.join(cluster_path, img_name), face_data["crop"])

        meta_file = os.path.join(cluster_path, "metadata.json")
        meta = json.load(open(meta_file)) if os.path.exists(meta_file) else []
        meta.append({
            "image": img_name,
            "source_type": face_data.get("source_type"),
            "source_name": face_data.get("source_name"),
            "timestamp_sec": face_data.get("timestamp_sec"),
        })
        with open(meta_file, "w") as f:
            json.dump(meta, f, indent=2)

        return target_cluster

    def promote_cluster(self, cluster_id: str, person_name: str):
        """Moves labeled face crops from the unknown queue into known_faces/<person_name>."""
        src_dir = os.path.join(self.queue_dir, cluster_id)
        if not os.path.exists(src_dir):
            return

        clean_name = person_name.strip().replace(" ", "_")
        dest_dir = os.path.join(self.known_dir, clean_name)
        os.makedirs(dest_dir, exist_ok=True)

        for fname in os.listdir(src_dir):
            if fname.endswith((".jpg", ".png", ".jpeg")):
                shutil.move(os.path.join(src_dir, fname), os.path.join(dest_dir, fname))

        shutil.rmtree(src_dir)


queue_manager = UnknownQueueManager(
    queue_dir=QUEUE_DIR,
    known_dir=KNOWN_DIR,
    cluster_threshold=CLUSTER_THRESHOLD,
    duplicate_threshold=DUPLICATE_THRESHOLD,
)
print(f"Queue manager ready (cluster_thresh={CLUSTER_THRESHOLD}, dup_thresh={DUPLICATE_THRESHOLD}) ✅")
````

## 6. Recognition Logger

Appends every inference result (recognized or queued) to a CSV log with timestamps.

````python
class RecognitionLogger:
    def __init__(self, log_path="data/logs/recognition_logs.csv"):
        self.log_path = log_path
        os.makedirs(os.path.dirname(self.log_path), exist_ok=True)
        self.columns = [
            "logged_at", "source_type", "source_name", "timestamp_sec",
            "predicted_label", "similarity_score", "detection_score", "queue_status",
        ]
        if not os.path.exists(self.log_path):
            pd.DataFrame(columns=self.columns).to_csv(self.log_path, index=False)

    def log_batch(self, records: list[dict]):
        """Appends a batch of inference results to the CSV log file."""
        if not records:
            return
        now_str = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        rows = []
        for r in records:
            rows.append({
                "logged_at": now_str,
                "source_type": r.get("source_type"),
                "source_name": r.get("source_name"),
                "timestamp_sec": r.get("timestamp_sec") if r.get("timestamp_sec") is not None else "",
                "predicted_label": r.get("pred_label"),
                "similarity_score": round(float(r.get("pred_score", 0.0)), 4),
                "detection_score": round(float(r.get("det_score", 0.0)), 4),
                "queue_status": r.get("queue_status", "recognized"),
            })
        df = pd.DataFrame(rows, columns=self.columns)
        df.to_csv(self.log_path, mode="a", header=False, index=False)

    def load_logs(self) -> pd.DataFrame:
        if os.path.exists(self.log_path):
            return pd.read_csv(self.log_path)
        return pd.DataFrame(columns=self.columns)


logger = RecognitionLogger(log_path=LOG_PATH)
print("Logger ready ✅")
````

## 7. Training

Scans `known_faces/`, extracts embeddings, and trains:
- **SVM** (linear, with probabilities) if ≥ 2 classes
- **KNN** (k=1, cosine) if only 1 class

````python
def train_classifier(
    known_dir=KNOWN_DIR,
    model_path=MODEL_PATH,
    embeddings_path=EMBEDDINGS_PATH,
    detector_instance=None,
):
    print("Starting training...")

    det = detector_instance or detector
    X, y = [], []

    for person_name in sorted(os.listdir(known_dir)):
        person_dir = os.path.join(known_dir, person_name)
        if not os.path.isdir(person_dir):
            continue

        for img_name in os.listdir(person_dir):
            if not img_name.lower().endswith((".jpg", ".jpeg", ".png", ".webp")):
                continue

            img_path = os.path.join(person_dir, img_name)
            frame = cv2.imread(img_path)
            if frame is None:
                continue

            faces = det.app.get(frame)

            # Pad tight crops so the detector can find the face
            if len(faces) == 0:
                h, w, _ = frame.shape
                pad_y, pad_x = int(h * 0.5), int(w * 0.5)
                padded = cv2.copyMakeBorder(
                    frame, pad_y, pad_y, pad_x, pad_x, cv2.BORDER_CONSTANT, value=[0, 0, 0]
                )
                faces = det.app.get(padded)

            if len(faces) > 0:
                largest = max(faces, key=lambda f: (f.bbox[2] - f.bbox[0]) * (f.bbox[3] - f.bbox[1]))
                X.append(largest.normed_embedding)
                y.append(person_name)

    if len(X) == 0:
        return {"status": "empty", "message": "No valid face embeddings found in known_faces."}

    X = np.array(X)
    y = np.array(y)
    unique_classes = np.unique(y)

    if len(unique_classes) >= 2:
        print("Using Support Vector Classifier")
        clf = SVC(kernel="linear", probability=True, C=1.0)
        clf.fit(X, y)
        model_type = "svm"
    else:
        print("Using KNN")
        clf = KNeighborsClassifier(n_neighbors=1, metric="cosine")
        clf.fit(X, y)
        model_type = "knn"

    os.makedirs(os.path.dirname(model_path), exist_ok=True)
    os.makedirs(os.path.dirname(embeddings_path), exist_ok=True)

    artifact = {
        "classifier": clf,
        "model_type": model_type,
        "classes": unique_classes.tolist(),
        "embeddings": X,
        "labels": y,
    }
    joblib.dump(artifact, model_path)
    joblib.dump({"X": X, "y": y}, embeddings_path)

    result = {
        "status": "success",
        "num_samples": len(X),
        "classes": unique_classes.tolist(),
        "model_type": model_type,
    }
    print(f"Training complete: {result}")
    return result
````

### 7.1 Run Training

Execute the cell below to (re)train the classifier on the current `known_faces/` dataset.

````python
train_result = train_classifier()
train_result
````

## 8. Inference

Predict the identity of faces in new images or a video.

````python
def predict_face(embedding: np.ndarray, artifact: dict, sim_threshold: float):
    """Matches an embedding against known faces using cosine similarity."""
    if artifact is None:
        return "Unknown", 0.0

    known_embs = artifact["embeddings"]
    known_labels = artifact["labels"]

    sims = cosine_similarity(embedding.reshape(1, -1), known_embs)
    best_idx = int(np.argmax(sims))
    best_sim = float(sims[0, best_idx])

    if best_sim < sim_threshold:
        return "Unknown", best_sim

    if artifact["model_type"] == "svm":
        pred = artifact["classifier"].predict(embedding.reshape(1, -1))
        return str(pred[0]), best_sim

    return str(known_labels[best_idx]), best_sim


def load_model(model_path=MODEL_PATH):
    if os.path.exists(model_path):
        return joblib.load(model_path)
    return None
````

### 8.1 Run Inference on Images

Point `IMAGE_PATHS` at the images you want to process, then run the cell.

````python
# ── Set your input image paths here ──
IMAGE_PATHS = []  # e.g. ["path/to/img1.jpg", "path/to/img2.jpg"]

if IMAGE_PATHS:
    artifact = load_model()
    faces = ingestor.process_images(IMAGE_PATHS)
    print(f"Detected {len(faces)} face(s)")

    for face in faces:
        label, score = predict_face(face["embedding"], artifact, SIM_THRESHOLD)
        face["pred_label"] = label
        face["pred_score"] = score

        if label == "Unknown":
            status = queue_manager.add_unknown_face(face)
            face["queue_status"] = status
        else:
            face["queue_status"] = "recognized"

        # Display the face crop
        rgb = cv2.cvtColor(face["crop"], cv2.COLOR_BGR2RGB)
        plt.figure(figsize=(2, 2))
        plt.imshow(rgb)
        plt.title(f"{label} ({score:.2f})")
        plt.axis("off")
        plt.show()

    logger.log_batch(faces)
    print("Inference complete and logged.")
else:
    print("⚠️  Set IMAGE_PATHS above to run inference on images.")
````

### 8.2 Run Inference on a Video

Set `VIDEO_PATH` below to process a video file.

````python
# ── Set your video path here ──
VIDEO_PATH = ""  # e.g. "path/to/clip.mp4"

if VIDEO_PATH:
    artifact = load_model()
    faces = ingestor.process_video(VIDEO_PATH)
    print(f"Extracted {len(faces)} face(s) from video")

    recognized, queued = 0, 0
    for face in faces:
        label, score = predict_face(face["embedding"], artifact, SIM_THRESHOLD)
        face["pred_label"] = label
        face["pred_score"] = score

        if label == "Unknown":
            status = queue_manager.add_unknown_face(face)
            face["queue_status"] = status
            if not status.startswith("skipped_duplicate"):
                queued += 1
        else:
            face["queue_status"] = "recognized"
            recognized += 1

    logger.log_batch(faces)
    print(f"Results: {recognized} recognized, {queued} new unknowns queued")
else:
    print("⚠️  Set VIDEO_PATH above to run inference on a video.")
````

## 9. Human-in-the-Loop Labeling

Browse unknown clusters and label them. The crops are promoted to `known_faces/` and the model is retrained.

````python
def show_unknown_clusters(queue_dir=QUEUE_DIR):
    """Displays all current unknown clusters with thumbnails."""
    if not os.path.exists(queue_dir):
        print("Queue is empty.")
        return []

    clusters = sorted([
        d for d in os.listdir(queue_dir)
        if os.path.isdir(os.path.join(queue_dir, d))
    ])

    if not clusters:
        print("✅ Queue is empty — all faces have been labeled or recognized.")
        return []

    for cid in clusters:
        cpath = os.path.join(queue_dir, cid)
        imgs = sorted([f for f in os.listdir(cpath) if f.endswith((".jpg", ".png", ".jpeg"))])
        if not imgs:
            continue

        print(f"\n{'='*50}")
        print(f"Cluster: {cid}  ({len(imgs)} samples)")
        print(f"{'='*50}")

        fig, axes = plt.subplots(1, min(len(imgs), 6), figsize=(12, 2.5))
        if min(len(imgs), 6) == 1:
            axes = [axes]
        for i, img_name in enumerate(imgs[:6]):
            bgr = cv2.imread(os.path.join(cpath, img_name))
            if bgr is not None:
                axes[i].imshow(cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB))
                axes[i].axis("off")
        plt.tight_layout()
        plt.show()

    return clusters


clusters = show_unknown_clusters()
````

````python
# ── Label a cluster ──────────────────────────────────────
# Set CLUSTER_ID to one of the cluster IDs shown above, and PERSON_NAME to the label.

CLUSTER_ID  = ""   # e.g. "person_443402"
PERSON_NAME = ""   # e.g. "Alice"

if CLUSTER_ID and PERSON_NAME:
    queue_manager.promote_cluster(CLUSTER_ID, PERSON_NAME)
    print(f"Promoted cluster '{CLUSTER_ID}' → known_faces/{PERSON_NAME}")
    print("Retraining model...")
    train_classifier()
    print("Done! ✅")
else:
    print("⚠️  Set CLUSTER_ID and PERSON_NAME above to label a cluster.")
````

## 10. Known Dataset Overview

Visualize the current known identities and their reference images.

````python
def show_known_dataset(known_dir=KNOWN_DIR):
    if not os.path.exists(known_dir):
        print("No known identities yet.")
        return

    people = sorted([d for d in os.listdir(known_dir) if os.path.isdir(os.path.join(known_dir, d))])

    if not people:
        print("No known identities yet.")
        return

    for person in people:
        pdir = os.path.join(known_dir, person)
        pimgs = [f for f in os.listdir(pdir) if f.endswith((".jpg", ".jpeg", ".png", ".webp"))]
        print(f"\n{person} ({len(pimgs)} images)")

        if pimgs:
            n = min(len(pimgs), 6)
            fig, axes = plt.subplots(1, n, figsize=(2.5 * n, 2.5))
            if n == 1:
                axes = [axes]
            for idx, img_file in enumerate(pimgs[:6]):
                bgr = cv2.imread(os.path.join(pdir, img_file))
                if bgr is not None:
                    axes[idx].imshow(cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB))
                    axes[idx].set_title(img_file[:12], fontsize=8)
                    axes[idx].axis("off")
            plt.tight_layout()
            plt.show()


show_known_dataset()
````

## 11. Evaluation

Measure model performance against a held-out test set.

**Expected test set structure:**
```
data/test_set/
  ├── PersonA/
  │   ├── img1.jpg
  │   └── img2.jpg
  ├── PersonB/
  │   └── img3.jpg
  └── Unknown/
      ├── stranger1.jpg
      └── stranger2.jpg
```

````python
def generate_evaluation_plots(y_true, y_pred, best_sims, sim_threshold, plot_dir=PLOT_DIR):
    os.makedirs(plot_dir, exist_ok=True)
    labels = sorted(list(set(y_true) | set(y_pred)))
    saved_paths = {}

    # 1. Confusion Matrix
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    fig, ax = plt.subplots(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=labels, yticklabels=labels, cbar=False, ax=ax)
    ax.set_title("Face Recognition Confusion Matrix")
    ax.set_xlabel("Predicted Identity")
    ax.set_ylabel("Actual Identity")
    plt.tight_layout()
    cm_path = os.path.join(plot_dir, "confusion_matrix.png")
    fig.savefig(cm_path, dpi=300)
    saved_paths["confusion_matrix"] = cm_path
    plt.show()

    # 2. Cosine Similarity Distribution
    fig, ax = plt.subplots(figsize=(7, 4))
    known_sims = [s for s, t in zip(best_sims, y_true) if t.lower() != "unknown"]
    unknown_sims = [s for s, t in zip(best_sims, y_true) if t.lower() == "unknown"]
    if known_sims:
        sns.histplot(known_sims, bins=15, kde=len(known_sims) > 1, color="#2b8cbe",
                     label="Known Persons", alpha=0.6, ax=ax)
    if unknown_sims:
        sns.histplot(unknown_sims, bins=15, kde=len(unknown_sims) > 1, color="#e34a33",
                     label="Unknown / Strangers", alpha=0.6, ax=ax)
    ax.axvline(sim_threshold, color="black", linestyle="--", linewidth=1.5,
               label=f"Threshold ({sim_threshold})")
    ax.set_title("Max Cosine Similarity Distribution")
    ax.set_xlabel("Cosine Similarity Score")
    ax.set_ylabel("Image Count")
    ax.set_xlim(0, 1.0)
    ax.legend()
    plt.tight_layout()
    sim_path = os.path.join(plot_dir, "similarity_distribution.png")
    fig.savefig(sim_path, dpi=300)
    saved_paths["similarity_distribution"] = sim_path
    plt.show()

    # 3. Per-Class Metrics Bar Chart
    prec, rec, f1, _ = precision_recall_fscore_support(y_true, y_pred, labels=labels, zero_division=0)
    x = np.arange(len(labels))
    width = 0.25

    fig, ax = plt.subplots(figsize=(7, 4))
    ax.bar(x - width, prec, width, label="Precision", color="#1b9e77")
    ax.bar(x, rec, width, label="Recall", color="#d95f02")
    ax.bar(x + width, f1, width, label="F1-Score", color="#7570b3")
    ax.set_xticks(x)
    ax.set_xticklabels(labels, rotation=15)
    ax.set_ylim(0, 1.1)
    ax.set_ylabel("Score")
    ax.set_title("Per-Class Evaluation Metrics")
    ax.legend(loc="lower right")
    plt.tight_layout()
    metrics_path = os.path.join(plot_dir, "class_metrics.png")
    fig.savefig(metrics_path, dpi=300)
    saved_paths["class_metrics"] = metrics_path
    plt.show()

    return saved_paths
````

````python
def evaluate_model(
    test_dir=TEST_DIR,
    model_path=MODEL_PATH,
    sim_threshold=SIM_THRESHOLD,
    detector_instance=None,
    plot_dir=PLOT_DIR,
):
    if not os.path.exists(model_path):
        return {"status": "error", "message": "No trained model found. Train the model first."}

    if not os.path.exists(test_dir) or not os.listdir(test_dir):
        return {
            "status": "error",
            "message": f"Test directory '{test_dir}' is empty. Add subfolders per person (and 'Unknown' for strangers).",
        }

    det = detector_instance or detector

    artifact = joblib.load(model_path)
    known_embs = artifact["embeddings"]
    known_labels = artifact["labels"]
    clf = artifact["classifier"]
    model_type = artifact["model_type"]

    y_true, y_pred, best_sims = [], [], []

    for folder_name in sorted(os.listdir(test_dir)):
        folder_path = os.path.join(test_dir, folder_name)
        if not os.path.isdir(folder_path):
            continue

        for img_name in os.listdir(folder_path):
            if not img_name.lower().endswith((".jpg", ".jpeg", ".png", ".webp")):
                continue

            frame = cv2.imread(os.path.join(folder_path, img_name))
            if frame is None:
                continue

            faces = det.app.get(frame)
            if not faces:
                continue

            largest = max(faces, key=lambda f: (f.bbox[2] - f.bbox[0]) * (f.bbox[3] - f.bbox[1]))
            emb = largest.normed_embedding.reshape(1, -1)

            sims = cosine_similarity(emb, known_embs)[0]
            best_idx = int(np.argmax(sims))
            best_sim = float(sims[best_idx])

            if best_sim < sim_threshold:
                pred = "Unknown"
            elif model_type == "svm":
                pred = str(clf.predict(emb)[0])
            else:
                pred = str(known_labels[best_idx])

            y_true.append(folder_name)
            y_pred.append(pred)
            best_sims.append(best_sim)

    if not y_true:
        return {"status": "error", "message": "No detectable faces found in the test set images."}

    acc = accuracy_score(y_true, y_pred)
    precision, recall, f1, _ = precision_recall_fscore_support(
        y_true, y_pred, average="weighted", zero_division=0
    )

    # Unknown Rejection Rate & False Acceptance Rate
    unknown_indices = [i for i, label in enumerate(y_true) if label.lower() == "unknown"]
    if unknown_indices:
        correctly_rejected = sum(1 for i in unknown_indices if y_pred[i].lower() == "unknown")
        unknown_rejection_rate = correctly_rejected / len(unknown_indices)
        false_acceptance_rate = 1.0 - unknown_rejection_rate
    else:
        unknown_rejection_rate = None
        false_acceptance_rate = None

    report_text = classification_report(y_true, y_pred, zero_division=0)
    plot_paths = generate_evaluation_plots(y_true, y_pred, best_sims, sim_threshold, plot_dir=plot_dir)

    metrics = {
        "status": "success",
        "total_test_samples": len(y_true),
        "sim_threshold": sim_threshold,
        "accuracy": round(float(acc), 4),
        "precision": round(float(precision), 4),
        "recall": round(float(recall), 4),
        "f1_score": round(float(f1), 4),
        "unknown_rejection_rate": round(float(unknown_rejection_rate), 4) if unknown_rejection_rate is not None else "N/A (No Unknown folder)",
        "false_acceptance_rate": round(float(false_acceptance_rate), 4) if false_acceptance_rate is not None else "N/A",
        "classification_report": report_text,
        "plots": plot_paths,
    }

    os.makedirs("data/logs", exist_ok=True)
    with open("data/logs/latest_evaluation.json", "w") as f:
        json.dump(metrics, f, indent=2)

    return metrics
````

### 11.1 Run Evaluation

````python
eval_result = evaluate_model()

if eval_result["status"] == "success":
    print(f"\n{'='*50}")
    print(f"  EVALUATION RESULTS")
    print(f"{'='*50}")
    print(f"  Total Test Samples : {eval_result['total_test_samples']}")
    print(f"  Similarity Thresh. : {eval_result['sim_threshold']}")
    print(f"  Accuracy           : {eval_result['accuracy'] * 100:.1f}%")
    print(f"  Precision (wt.)    : {eval_result['precision'] * 100:.1f}%")
    print(f"  Recall (wt.)       : {eval_result['recall'] * 100:.1f}%")
    print(f"  F1 Score (wt.)     : {eval_result['f1_score'] * 100:.1f}%")
    print(f"  Unknown Rej. Rate  : {eval_result['unknown_rejection_rate']}")
    print(f"  False Accept. Rate : {eval_result['false_acceptance_rate']}")
    print(f"{'='*50}")
    print(f"\nPer-Class Classification Report:")
    print(eval_result["classification_report"])
else:
    print(f"⚠️  {eval_result['message']}")
````
